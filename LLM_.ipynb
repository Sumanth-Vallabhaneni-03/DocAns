```python
# ===============================
# Install dependencies
# ===============================
!apt-get update -y
!apt-get install -y zstd
!curl -fsSL https://ollama.com/install.sh | sh
!pip install pyngrok requests -q


# ===============================
# ENTER YOUR NGROK TOKEN
# ===============================
NGROK_AUTH_TOKEN = "3BzboDbsKzlpPEQaZHvwZtjEvTC_7JeCKdWp7bV8j6fdTjkv4"


# ===============================
# Import libraries
# ===============================
import os
import threading
import subprocess
import time
import requests
from pyngrok import ngrok


# ===============================
# Authenticate ngrok
# ===============================
ngrok.set_auth_token(NGROK_AUTH_TOKEN)


# ===============================
# Allow external connections
# ===============================
os.environ["OLLAMA_HOST"] = "0.0.0.0:11434"
os.environ["OLLAMA_ORIGINS"] = "*"


# ===============================
# Start Ollama server
# ===============================
def run_ollama():
    subprocess.run(["ollama", "serve"])

threading.Thread(target=run_ollama, daemon=True).start()


# ===============================
# Wait for server startup
# ===============================
print("Starting Ollama server...")
time.sleep(10)


# ===============================
# Start ngrok tunnel
# ===============================
public_url = ngrok.connect(11434).public_url
print("\nPUBLIC OLLAMA API URL:")
print(public_url)


# ===============================
# Download models
# ===============================
print("\nDownloading models (first run may take a few minutes)...")

subprocess.run(["ollama","pull","qwen3:8b-q4_K_M"])
subprocess.run(["ollama","pull","nomic-embed-text"])


print("\nModels installed successfully!")


# ===============================
# Test the LLM
# ===============================
print("\nTesting Qwen3 model...")

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "qwen3:8b-q4_K_M",
        "prompt": "Explain artificial intelligence in simple terms.",
        "stream": False
    }
)

print("\nModel response:")
print(response.json()["response"])


# ===============================
# Test embeddings model
# ===============================
print("\nTesting embedding model...")

embed = requests.post(
    "http://localhost:11434/api/embeddings",
    json={
        "model": "nomic-embed-text",
        "prompt": "Artificial intelligence is transforming technology"
    }
)

print("Embedding vector length:", len(embed.json()["embedding"]))


# ===============================
# Keep Colab alive
# ===============================
print("\nServer running. Keep this notebook open.")

while True:
    time.sleep(60)
```
